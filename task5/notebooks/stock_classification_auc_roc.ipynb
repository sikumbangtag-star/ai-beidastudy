## 1. ?????????????

```python
from IPython.display import Image, display, HTML

from task5.src.stock_classification import (
    DATA_PATH,
    OUTPUT_DIR,
    FIGURE_DIR,
    MODEL_DISPLAY_NAMES,
    MODEL_NAMES,
    load_raw_data,
    prepare_dataset,
    split_by_date_order,
    build_feature_frame,
    fit_all_models,
    save_results,
)

raw = load_raw_data(DATA_PATH)
print('raw shape:', raw.shape)
print(raw['Y'].value_counts(dropna=False))
print('positive_rate =', round(raw['Y'].mean(), 4))
raw.head()
```

## 2. ???????????????

????????? 70% ??????? 20% ??????? 10% ????????????????????

```python
prepared = prepare_dataset(raw)
train, valid, test = split_by_date_order(prepared)
features = build_feature_frame(train, valid, test)

for name, part in [('train', train), ('valid', valid), ('test', test)]:
    print(name, part.shape, part['Date'].min(), part['Date'].max(), 'positive_rate=', round(part['Y'].mean(), 4))
print('feature_count =', len(features.columns))
```

## 3. ????????? AUC

????????????????????????????????????????????? `[0, 1]` ??? AUC/ROC ???

```python
result = fit_all_models(features)
metrics = result['metrics']
predictions = result['predictions']
metrics
```

## 4. ??????? HTML ??

????????????????????EDA payload???? ROC ?? HTML ?????

```python
save_results(metrics, predictions, result['feature_importance'])

from task5.src.stock_classification import build_eda_payload
import json
payload = build_eda_payload(prepared, features, result)
(OUTPUT_DIR / 'eda_model_payload.json').write_text(json.dumps(payload, ensure_ascii=False, indent=2), encoding='utf-8')

import task5.scripts.build_ml_panel as panel
panel.main()
print('HTML panel:', 'task5/design/ml_classification_panel.html')
```
